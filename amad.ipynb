```python
import requests
import json

BASE_URL = "http://localhost:1234/v1"

# 1. Check models loaded in LM Studio
models_response = requests.get(f"{BASE_URL}/models")

print("Models status:", models_response.status_code)
print(models_response.text)

models_data = models_response.json()
MODEL_NAME = models_data["data"][0]["id"]

print("Using model:", MODEL_NAME)
```

```python
import requests
import json

BASE_URL = "http://localhost:1234/v1"
API_URL = f"{BASE_URL}/chat/completions"

messages = [
   
    {
        "role": "user",
        "content": "Describe different video formats in video."
    }
]

payload = {
    "model": MODEL_NAME,
    "messages": messages,
    "temperature": 0.7,
    "max_tokens": 300,
    "stream": False
}

response = requests.post(API_URL, json=payload)

print("Status:", response.status_code)
print("Raw response:")
print(response.text)

if response.status_code == 200:
    data = response.json()
    print("\nAI Answer:")
    print(data["choices"][0]["message"]["content"])
else:
    print("\nLM Studio returned an error.")
```
